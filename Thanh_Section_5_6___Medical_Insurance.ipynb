##2.0. Dataset description
The dataset represents health insurance information for a sample of individuals, containing various features that might affect their insurance charges.

The data includes demographic details, health-related factors, and the corresponding insurance charges billed to each individual.

The primary purpose of this dataset is to explore the relationships between these factors and the insurance charges, which can be useful for predictive modelling and understanding the cost drivers in health insurance.

**Columns:**

1.age: The age of the individual in years.

2.sex: The gender of the individual (male or female).

3.bmi: Body Mass Index (BMI) of the individual, which is a measure of body fat based on height and weight.

4.children: The number of children or dependents covered by the insurance.

5.smoker: Indicates whether the individual is a smoker (yes or no).

6.region: The residential area of the individual in the United States, divided into four regions: northeast, northwest, southeast, southwest.

7.charges: The medical insurance costs billed to the individual.

##2.1. Load The Data

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```
# Load the data by link
df = pd.read_csv('https://raw.githubusercontent.com/sca-programming-school/datasets/main/insurance.csv')
```

##2.2. Convert Categorical data to Nummeric

```
df.head()
```

```
df.info()
```

```
#Use One_hot encoding to convert categorical data to numeric
df_encoded = pd.get_dummies(df, columns=['sex', 'smoker', 'region'])
```

```
#Display the first few row of the encoded dataframe
print(df_encoded.head())
```

##2.3. Build a Regression Model

```
# Convert all columns to numeric dtype, if possible
df_encoded = df_encoded.apply(pd.to_numeric, errors='coerce')
```

```
# Drop any rows with NaN values after conversion
df_encoded = df_encoded.dropna()
```

```
# Define the independent variables (X) and the dependent variable (y)
X = df_encoded.drop('charges', axis=1)
y = df_encoded['charges']
```

```
# Add a constant to the model (intercept)
X = sm.add_constant(X)
```

```
# Build the regression model
model = sm.OLS(y, X).fit()
```

```
# Output the model summary
print(model.summary())
```
